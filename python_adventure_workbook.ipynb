## 3. Numbers and Math

Python can be a super calculator. It uses `+` to add, `-` to subtract, `*` to multiply, and `/` to divide. Parentheses work just like they do in math class.

````python
print(3 + 4)
print(10 - 2)
print(5 * 3)
print(12 / 4)
print((2 + 3) * 4)
````

There are two common kinds of numbers:

- `int` means a whole number, such as `7`.
- `float` means a number with a decimal, such as `2.5`.

`**` means “to the power of.” For example, `3 ** 2` is 3 × 3.

````python
cookies = 30
friends = 5
cookies_each = cookies / friends

print("Each friend gets", cookies_each, "cookies.")
print("A 4 by 4 square has", 4 ** 2, "small squares.")
````

## 7. Making Choices With if

An `if` statement lets a program choose. The indented line runs only when the question is `True`. `else` tells Python what to do otherwise.

````python
weather = "rainy"

if weather == "rainy":
    print("Take an umbrella!")
else:
    print("Wear your sunglasses!")
````

The colon `:` and the four spaces before the indented lines matter. Indentation shows which instructions belong inside the choice. `elif` means “otherwise, if...”

````python
score = 7

if score >= 10:
    print("Gold star!")
elif score >= 5:
    print("Great job!")
else:
    print("Keep trying—you are learning!")
````

## 8. Loops — Repeat Without Re-typing

A **loop** repeats instructions. A `for` loop is useful when we know how many times to repeat. `range(5)` gives the numbers 0, 1, 2, 3, and 4.

````python
for number in range(5):
    print("Blastoff number:", number)
````

````python
for countdown in range(5, 0, -1):
    print(countdown)
print("🚀 Blast off!")
````

The name after `for` is a variable. It changes each trip around the loop. The indented instructions are the ones that repeat.

## 10. Functions — Make Your Own Commands

A **function** is a named group of instructions. Build one with `def`, then call it by writing its name with parentheses.

````python
def do_robot_dance():
    print("Step left!")
    print("Beep beep!")
    print("Spin around!")

do_robot_dance()
````

A function can receive information called a **parameter**. Here, `name` is the parameter. Each call can send a different name.

````python
def greet(name):
    print(f"Hello, {name}! Welcome to Python Land.")

greet("Ari")
greet("Bo")
````

A function can also `return` an answer. Returning sends a value back so the rest of the program can use it.

````python
def double(number):
    return number * 2

result = double(6)
print("Double 6 is", result)
````

## 11. Bugs and Debugging

A **bug** is a mistake in code. **Debugging** means finding and fixing it. Bugs do not mean you are bad at coding—they mean you are coding!

When Python shows an error, read the last line first. It often tells you the type of problem and where to look.

Common bugs:

- missing quotation mark or parenthesis
- misspelled variable name
- missing colon after `if`, `for`, or `def`
- indentation that does not line up

## Mini-Project 2 — Pocket Quiz

This quiz uses variables, comparisons, `if`, and a score. Change the pretend answers to test it. Later, you can use `input()` to ask a real player.

````python
score = 0

# Pretend answers make the notebook run without pausing.
answer_1 = "8"
answer_2 = "blue"

if answer_1 == "8":
    print("Question 1: Correct!")
    score = score + 1
else:
    print("Question 1: Nice try. 5 + 3 is 8.")

if answer_2.lower() == "blue":
    print("Question 2: Correct!")
    score = score + 1
else:
    print("Question 2: Nice try. Our expected answer was blue.")

print(f"Final score: {score} out of 2")

# Interactive upgrade:
# answer_1 = input("What is 5 + 3? ")
# answer_2 = input("Name a color of the daytime sky: ")
````

## Mini-Project 3 — Tiny Pet Adventure

This program combines a function, a list, a loop, and choices. Change the pet, energy, and backpack to create new adventures.

````python
pet_name = "Biscuit"
energy = 7
backpack = ["map", "apple", "tiny flashlight"]

def show_backpack(items):
    print("Backpack:")
    for item in items:
        print("-", item)

print(f"{pet_name} begins an adventure!")
show_backpack(backpack)

if energy >= 5:
    print(f"{pet_name} has enough energy to climb Sparkle Hill.")
    energy = energy - 3
else:
    print(f"{pet_name} rests under a tree.")

print("Energy left:", energy)
````
